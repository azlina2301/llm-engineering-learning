```python
import os
import httpx
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
MODELS = {
    "gpt": "gpt-5-mini",
    "google": "gemini-3.5-flash",
    "alibaba": "qwen3.6-27b"
}


openai = OpenAI(
    base_url=os.getenv('LITELLM_BASE_URL'),
    api_key=os.getenv('LITELLM_API_KEY'),
    http_client=httpx.Client(verify=False, timeout=60.0)
    )


class ChatBot:

    def __init__(self):
        self.chat_history = []


    def chat(self, model_type, user_chat):

        if model_type not in MODELS:
            raise KeyError (f"Uknown Model: {model_type}, choose one of it: {MODELS}")
             
        message = [
            {"role": "system", "content": "You are very helpful assistant that always provide the informations with concise, precise and short"},
            ]
        
        self.chat_history.append(
            {
                "role": "user", 
                "content": user_chat
                }
            )
        
        response = openai.chat.completions.create(
            model=MODELS[model_type], 
            messages=message + self.chat_history
            )
        
        return display(Markdown(response.choices[0].message.content))
```

```python
chatbot = ChatBot()

chatbot.chat("gpt", "Hi")
```

```python
chatbot.chat("gpt", "I am Duck, I live in Bosnia. I am having problem with focus whenever I read the book.")
```

```python
chatbot.chat("gpt", "Do you know What I was talking before?")
```

```python
chatbot.chat("google", "How can you help me in this case?")
```
